```python
from huggingface_hub import login
import os
import datasets
login(token="") # token=Your_HuggingFace_Token
# load metadata
import pandas as pd
meta_df = pd.read_csv("./HEST_v1_1_0.csv") #For details, please refer to the 'Query HEST by organ, technology, oncotree code' section on the HEST Hugging Face main page.
```

# Dataset - Human Kidney

```python
data_path = "./hest1k_datasets/kidney"

ids_to_query = meta_df.loc[
    meta_df["dataset_title"].str.startswith(
        "Spatial localization with Spatial Transcriptomics for an atlas of healthy and injured cell states"
    ),
    "id"
].tolist()

print(len(ids_to_query))
print(ids_to_query)

list_patterns = []
for sid in ids_to_query:
    list_patterns.extend([
        f"wsis/{sid}*",
        f"st/{sid}*",
        f"metadata/{sid}*",
        f"tissue_seg/{sid}*",
    ])

dataset = datasets.load_dataset(
    "MahmoodLab/hest",
    cache_dir=data_path,
    patterns=list_patterns,
    trust_remote_code=True
)
```
